# TauGrid CPU Ray demo

Submitted through the TauGrid JupyterLab plugin. It runs on a CPU-only Ray worker
profile, so it needs no GPU quota.


```python
import time
from math import sqrt

import ray

# On the cluster the RayJob starts the cluster, so connect to it; locally, start one.
try:
    ray.init(address="auto", ignore_reinit_error=True, logging_level="ERROR")
except Exception:
    ray.init(ignore_reinit_error=True, logging_level="ERROR")

resources = ray.cluster_resources()
print("ray", ray.__version__)
print("nodes", len(ray.nodes()))
print("cpus", resources.get("CPU"))
print("gpus", resources.get("GPU", 0))
```

```python
@ray.remote(num_cpus=1)
def busy_sum(n: int) -> float:
    return sum(sqrt(i) for i in range(n))

started = time.time()
results = ray.get([busy_sum.remote(2_000_000) for _ in range(8)])
elapsed = time.time() - started
print(f"CPU_DEMO tasks={len(results)} total={sum(results):.2f} elapsed={elapsed:.2f}s")
```
