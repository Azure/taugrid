# Deterministic CPU loss curve
Pure Python gradient descent fits y = 2x + 1. Forty real observations span about forty seconds so the existing status poll can collect multiple windows. No downloads or GPU packages are needed. Submit using an explicitly reviewed CPU profile.


```python
import time

inputs = (-2.0, -1.0, 0.0, 1.0, 2.0)
targets = tuple(2.0 * value + 1.0 for value in inputs)
weight, bias = 0.0, 0.0
for step in range(40):
    errors = tuple(weight * value + bias - target for value, target in zip(inputs, targets))
    loss = sum(error * error for error in errors) / len(inputs)
    print(f'step={step} loss={loss:.12g}', flush=True)
    weight -= 0.1 * 2.0 * sum(error * value for error, value in zip(errors, inputs)) / len(inputs)
    bias -= 0.1 * 2.0 * sum(errors) / len(inputs)
    time.sleep(1.0)
print(f'Finished: weight={weight:.6f}, bias={bias:.6f}', flush=True)
```
