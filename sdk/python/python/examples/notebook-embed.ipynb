# TensorBoard-style TauGrid embed

`%taugrid --embed` serves the TauGrid portal run view through a local
same-origin proxy and frames it in this cell - the same pattern the
TensorBoard notebook plugin uses. Set `PORTAL` to your portal URL.


```python
%load_ext tau.widgets.ipython
```

```python
from IPython.display import display

PORTAL = 'http://127.0.0.1:8090'   # your TauGrid portal base URL
RUN = 'notebook-embed-demo'
NAMESPACE = 'tau-notebook-e2e'

view = get_ipython().run_line_magic(
    'taugrid',
    f'--embed --name={RUN} --namespace={NAMESPACE} --portal={PORTAL}',
)
print('proxied:', view.proxied)
print('iframe url:', view.url)
print('portal link:', view.portal_url + view.path)
display(view)
```
