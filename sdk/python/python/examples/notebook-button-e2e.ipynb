# TauGrid plugin button E2E

The panel below carries the real interactive Submit button. Its submit is
backed by an in-kernel recording fake, so a browser click can be asserted
end-to-end without a live cluster: the click drives the same handler the
button registers, and the status line repaints into run view.

```python
%load_ext tau.widgets.ipython
```

```python
from tau.widgets.panel import TauGridPanel
from tau._backend import SubmittedRun
from IPython.display import display

recorded = []

def fake_submit(notebook=None, **kwargs):
    recorded.append(notebook)
    panel.run_name = "submitted-button-demo"  # status repaint proves the click ran
    return SubmittedRun(name=panel.run_name, namespace=panel.namespace, kind="RayJob")

panel = TauGridPanel(namespace="ray")
panel.set_notebook_path("examples/notebook-button-e2e.ipynb")
panel.submit = fake_submit
display(panel.build())
```
